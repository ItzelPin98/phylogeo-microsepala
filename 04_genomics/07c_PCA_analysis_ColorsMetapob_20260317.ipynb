```python
#creamos el objeto PCA con los inputs y parametros necesarios
tool = ipa.pca(
    data=SNPS,
    imap=imap,
    minmap={i: 0 for i in imap},
    mincov=0.75,
    minmaf=0.02,
    # impute_method=12
)
```

```python
#corremos el pca
tool.run(100, subsample=True, seed=1245)
```

```python
# guardar los valores de PC en un dataframe
df = pd.DataFrame(tool.pcaxes[0], index=tool.names)

#guardar estos valores en un CSV
#df.to_csv("PCsmicro_clades_2026_0702.csv")

#ver los primeras 10 muestras y los primeros 10 ejes PC
df.iloc[:10, :10].round(2)
```

```python
vars = tool.variances[0]

pca_summary = pd.DataFrame({
    "PC": range(1, len(vars)+1),
    "Variance": vars,
    "Explained_%": (vars / vars.sum()) * 100
})

pca_summary.head(10)
```

```python
pca_summary.round(3)
pca_summary.round(3).to_csv("PCAgen_summary_20260702.txt", sep="\t", index=False)
```
